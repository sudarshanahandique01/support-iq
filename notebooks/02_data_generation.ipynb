Define the date range

```python
start_date = datetime(2025, 1, 1)

end_date = datetime(2026, 9, 1)
```

```python
def generate_random_date():

    days_difference = (
        end_date - start_date
    ).days

    random_days = random.randint(
        0,
        days_difference
    )

    random_date = (
        start_date
        + timedelta(days=random_days)
    )

    return random_date
```

Generate a closed date

```python
def generate_closed_date(created_date):

    resolution_days = random.randint(
        0,
        10
    )

    closed_date = (
        created_date
        + timedelta(days=resolution_days)
    )

    return closed_date
```

Generate ONE ticket

```python
def generate_ticket(ticket_number):

    # ----------------------------------------------
    # SELECT CATEGORY
    # ----------------------------------------------

    category = random.choice(
        categories
    )


    # ----------------------------------------------
    # SELECT SUBCATEGORY
    # based on category
    # ----------------------------------------------

    sub_category = random.choice(
        sub_categories[category]
    )


    # ----------------------------------------------
    # SELECT ISSUE
    # based on category + subcategory
    # ----------------------------------------------

    issue_description = random.choice(
        issue_descriptions[
            category
        ][
            sub_category
        ]
    )


    # ----------------------------------------------
    # GENERATE DATES
    # ----------------------------------------------

    created_date = generate_random_date()

    closed_date = generate_closed_date(
        created_date
    )


    # ----------------------------------------------
    # CREATE TICKET
    # ----------------------------------------------

    ticket = {

        "ticket_id":
            f"TKT{ticket_number:07d}",

        "customer_id":
            f"CUST{random.randint(1, 50000):05d}",

        "agent_id":
            f"AG{random.randint(1, 1000):04d}",

        "created_date":
            created_date,

        "closed_date":
            closed_date,

        "priority":
            random.choice(priorities),

        "category":
            category,

        "sub_category":
            sub_category,

        "product":
            "HCM",

        "status":
            random.choice(statuses),

        "channel":
            random.choice(channels),

        "issue_description":
            issue_description,

        "resolution":
    resolutions[
        category
    ][
        sub_category
    ],

        "customer_region":
            random.choice(regions),

        "first_response_minutes":
            random.randint(5, 600),

        "resolution_hours":
            round(
                random.uniform(1, 200),
                2
            ),

        "customer_satisfaction":
            random.randint(1, 5),

        "escalation_flag":
            random.choice(
                ["Yes", "No"]
            )
    }


    return ticket
```

Create the first dirty-data function — NULLs

```python
def introduce_nulls(df):

    df = df.copy()

    null_columns = [
        "customer_id",
        "agent_id",
        "priority",
        "resolution",
        "customer_satisfaction"
    ]

    for column in null_columns:

        number_of_nulls = int(
            len(df) * 0.05
        )

        random_indexes = random.sample(
            range(len(df)),
            number_of_nulls
        )

        df.loc[
            random_indexes,
            column
        ] = None

    return df
```

Create duplicate function

```python
def introduce_duplicates(df):

    df = df.copy()

    duplicate_count = int(
        len(df) * 0.02
    )

    duplicate_rows = df.sample(
        n=duplicate_count,
        random_state=42
    )

    df = pd.concat(
        [
            df,
            duplicate_rows
        ],
        ignore_index=True
    )

    return df
```

Add More Dirty Data

Invalid Priority

```python
def introduce_invalid_priority(df):

    df = df.copy()

    number_of_errors = int(
        len(df) * 0.02
    )

    random_indexes = random.sample(
        range(len(df)),
        number_of_errors
    )

    invalid_values = [
        "URGENT",
        "Super High",
        "Unknown Priority",
        "Very High"
    ]

    for index in random_indexes:

        df.loc[
            index,
            "priority"
        ] = random.choice(
            invalid_values
        )

    return df
```

```python
invalid_priority_df = introduce_invalid_priority(
    test_df
)

valid_priorities = [
    "Low",
    "Medium",
    "High",
    "Critical"
]

invalid_priority_values = invalid_priority_df[
    ~invalid_priority_df["priority"].isin(
        valid_priorities
    )
]["priority"]
```

Inconsistent Priority Formatting

```python
def introduce_inconsistent_priority(df):

    df = df.copy()

    number_of_errors = int(
        len(df) * 0.03
    )

    random_indexes = random.sample(
        range(len(df)),
        number_of_errors
    )

    inconsistent_values = [
        "high",
        "HIGH",
        " High",
        "high ",
        "critical",
        "CRITICAL",
        " low",
        "medium "
    ]

    for index in random_indexes:

        df.loc[
            index,
            "priority"
        ] = random.choice(
            inconsistent_values
        )

    return df
```

```python
def introduce_invalid_csat(df):

    df = df.copy()

    number_of_errors = int(
        len(df) * 0.02
    )

    random_indexes = random.sample(
        range(len(df)),
        number_of_errors
    )

    invalid_values = [
        0,
        6,
        7,
        10,
        -1
    ]

    for index in random_indexes:

        df.loc[
            index,
            "customer_satisfaction"
        ] = random.choice(
            invalid_values
        )

    return df
```

```python
def introduce_invalid_dates(df):

    df = df.copy()

    # Convert dates to strings
    df["created_date"] = df["created_date"].astype(str)

    number_of_errors = int(
        len(df) * 0.01
    )

    random_indexes = random.sample(
        range(len(df)),
        number_of_errors
    )

    invalid_dates = [
        "2026-02-30",
        "2026-02-31",
        "2026-13-01",
        "not_a_date",
        "2025-99-99"
    ]

    for index in random_indexes:

        df.loc[
            index,
            "created_date"
        ] = random.choice(
            invalid_dates
        )

    return df
```

Extreme Resolution Hours

```python
def introduce_outliers(df):

    df = df.copy()

    number_of_outliers = int(
        len(df) * 0.01
    )

    random_indexes = random.sample(
        range(len(df)),
        number_of_outliers
    )

    for index in random_indexes:

        df.loc[
            index,
            "resolution_hours"
        ] = random.uniform(
            5000,
            50000
        )

    return df
```

```python
outlier_df = introduce_outliers(
    test_df
)
```

```python
print(
    outlier_df[
        outlier_df["resolution_hours"] > 500
    ][
        [
            "ticket_id",
            "resolution_hours"
        ]
    ]
)
```

Combine All Dirty-Data Functions

```python
def create_dirty_dataset(df):

    df = introduce_nulls(df)

    df = introduce_duplicates(df)

    df = introduce_invalid_priority(df)

    df = introduce_inconsistent_priority(df)

    df = introduce_invalid_csat(df)

    df = introduce_invalid_dates(df)

    df = introduce_outliers(df)

    return df
```

Generate the Real 1.2 Million Dataset

```python
TOTAL_RECORDS = 1_200_000

CHUNK_SIZE = 50_000
```

Prepare the Output File

```python
import os

raw_path = "../data/raw/support_tickets_raw.csv"

os.makedirs(
    "../data/raw",
    exist_ok=True
)

if os.path.exists(raw_path):

    os.remove(raw_path)

print(
    "Ready to generate:",
    raw_path
)
```

Generate the Dataset

```python
first_chunk = True

for start in range(
    1,
    TOTAL_RECORDS + 1,
    CHUNK_SIZE
):

    end = min(
        start + CHUNK_SIZE,
        TOTAL_RECORDS + 1
    )

    records = []

    for ticket_number in range(
        start,
        end
    ):

        ticket = generate_ticket(
            ticket_number
        )

        records.append(ticket)

    chunk_df = pd.DataFrame(
        records
    )

    chunk_df = create_dirty_dataset(
        chunk_df
    )

    chunk_df.to_csv(
        raw_path,
        mode="w" if first_chunk else "a",
        header=first_chunk,
        index=False
    )

    first_chunk = False

    print(
        f"Generated records: "
        f"{start:,} - {end - 1:,}"
    )
```

```python
import pandas as pd

raw_df = pd.read_csv(
    raw_path
)

print(
    "Dataset shape:",
    raw_df.shape
)
```
