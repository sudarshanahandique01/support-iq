Import PySpark

```python
from pyspark.sql import SparkSession
```

Create SparkSession

```python
spark = (
    SparkSession.builder
    .appName("SupportIQ")
    .master("local[*]")
    .getOrCreate()
)
```

Define the Raw File Path

```python
raw_path = "../data/raw/support_tickets_raw.csv"
```

Read the CSV

```python
tickets_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(raw_path)
)
```

```python
print(
    "Data loaded successfully."
)

print(
    "Number of columns:",
    len(tickets_df.columns)
)
```

Display Column Names

```python
print(
    tickets_df.columns
)
```

Check the Schema

```python
tickets_df.printSchema()
```

Look at the First 10 Records

```python
tickets_df.show(10)
```

Check the Number of Rows

```python
row_count = tickets_df.count()

print(
    "Total rows:",
    row_count
)
```

Check the Shape


```python
print(
    "Rows:",
    tickets_df.count()
)

print(
    "Columns:",
    len(tickets_df.columns)
)
```

Check Distinct Priority Values

```python
tickets_df.select(
    "priority"
).distinct().show()
```

Check Distinct CSAT Values

```python
tickets_df.select(
    "customer_satisfaction"
).distinct().show()
```

Check Resolution Hours

```python
from pyspark.sql import functions as F
tickets_df.select(
    F.min("resolution_hours").alias("minimum"),
    F.max("resolution_hours").alias("maximum")
).show()
```

Check Spark DataFrame Partitions

```python
print(
    "Number of partitions:",
    tickets_df.rdd.getNumPartitions()
)
```
